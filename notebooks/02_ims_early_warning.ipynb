```python
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.stats import skew, kurtosis

# Project directories
ims_raw_dir = Path("../data/ims_raw")
ims_processed_dir = Path("../data/ims_processed")
results_dir = Path("../results")
figures_dir = Path("../figures")

# Create Phase II directories
ims_raw_dir.mkdir(parents=True, exist_ok=True)
ims_processed_dir.mkdir(parents=True, exist_ok=True)

print("IMS raw data:", ims_raw_dir.resolve())
print("IMS processed data:", ims_processed_dir.resolve())
```

```python
all_items = list(ims_raw_dir.rglob("*"))

files = [p for p in all_items if p.is_file()]
folders = [p for p in all_items if p.is_dir()]

print("Total folders:", len(folders))
print("Total files:", len(files))

print("\nTop-level contents:")
for item in sorted(ims_raw_dir.iterdir()):
    print(" ", item.name)

print("\nFirst 20 files found:")
for file in sorted(files)[:20]:
    print(" ", file.relative_to(ims_raw_dir))
```

```python
# Find all files whose names begin with the Test 2 year
test2_files = sorted([
    p for p in ims_raw_dir.rglob("*")
    if p.is_file() and p.name.startswith("2004.")
])

print("IMS Test 2 recordings found:", len(test2_files))

if len(test2_files) > 0:

    print("\nFirst 5 recordings:")
    for file in test2_files[:5]:
        print(" ", file.relative_to(ims_raw_dir))

    print("\nLast 5 recordings:")
    for file in test2_files[-5:]:
        print(" ", file.relative_to(ims_raw_dir))

    print("\nFirst recording:")
    print(test2_files[0].name)

    print("\nLast recording:")
    print(test2_files[-1].name)

else:
    print("\nNo extracted Test 2 recordings found.")
```
